```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import csv
import time

base_url = "https://books.toscrape.com/"

# إضافة User-Agent لتقليد متصفح حقيقي وتجنب الحظر
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36"
}

print("جاري جلب الصفحة الرئيسية لاستخراج التصنيفات...")
response = requests.get(base_url, headers=headers)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

# استخراج روابط التصنيفات
categories = soup.select(".side_categories ul li ul li a")

# القائمة التي ستجمع جميع المنتجات
scraped_data = []

ratings = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

# المرور على جميع التصنيفات
for category in categories:
    name = category.get_text(strip=True)
    category_url = urljoin(base_url, category["href"])

    print("=" * 50)
    print("التصنيف:", name)
    print("الرابط الأساسي:", category_url)

    current_page = 1

    # حلقة تكرار للتنقل بين صفحات التصنيف الواحد
    while True:
        print(f"  -> جاري جلب المنتجات من الصفحة {current_page}...")

        try:
            response = requests.get(category_url, headers=headers)
            response.raise_for_status()
        except requests.exceptions.RequestException as e:
            print(f"  [!] خطأ في جلب البيانات: {e}")
            break

        soup = BeautifulSoup(response.text, "html.parser")
        products_container = soup.find("ol", class_="row")

        if not products_container:
            break

        products = products_container.find_all("li", class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")

        for container in products:
            box = container.find("article", class_="product_pod")
            if not box:
                continue

            # استخراج البيانات
            product_name = box.select_one("h3 a")["title"]
            product_url = urljoin(base_url, box.select_one("h3 a")["href"])
            image_url = urljoin(base_url, box.select_one(".image_container img")["src"])
            price = box.select_one(".price_color").get_text(strip=True)

            rating_element = box.select_one("p.star-rating")
            rating_name = rating_element.get("class")[1] if rating_element and len(rating_element.get("class")) > 1 else "Zero"
            rating = ratings.get(rating_name, 0)

            availability = box.select_one(".availability").get_text(" ", strip=True)

            # إضافة البيانات للقائمة
            scraped_data.append({
                "التصنيف": name,
                "الاسم": product_name,
                "السعر": price,
                "التقييم": rating,
                "التوفر": availability,
                "الرابط": product_url,
                "الصورة": image_url
            })

        # التحقق من وجود صفحة تالية
        next_page = soup.select_one("li.next a")
        if next_page:
            current_page += 1
            category_url = urljoin(base_url, next_page["href"])
            time.sleep(1)  # تأخير بسيط لمدة ثانية لتجنب الضغط على الخادم
        else:
            print(f"  -> تم الانتهاء من جميع صفحات تصنيف: {name}")
            break

# حفظ البيانات في CSV
csv_filename = "data.csv"
with open(csv_filename, mode="w", encoding="utf-8-sig", newline="") as file:
    fieldnames = ["التصنيف", "الاسم", "السعر", "التقييم", "التوفر", "الرابط", "الصورة"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(scraped_data)

print("=" * 50)
print(f"✅ تم استخراج {len(scraped_data)} منتج بنجاح!")
print(f"💾 تم حفظ البيانات في الملف: {csv_filename}")
```

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import csv
import time

# 1. تعريف قائمة البيانات خارج الحلقة لتجميع كل الصفحات
scraped_data = []

ratings = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

# رابط الجذر الأساسي للموقع (أفضل للاستخدام مع urljoin)
root_url = "https://books.toscrape.com/"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36"
}

print("بدء عملية جلب البيانات من جميع الصفحات...")

for page_number in range(1, 51):
    # 2. تصحيح رابط الصفحة الأولى
    if page_number == 1:
        page_url = "https://books.toscrape.com/"
    else:
        page_url = f"https://books.toscrape.com/catalogue/page-{page_number}.html"

    print(f"جاري جلب المنتجات من الصفحة {page_number}...")

    try:
        response = requests.get(page_url, headers=headers)
        response.raise_for_status()
    except requests.exceptions.RequestException as e:
        print(f"  [!] فشل في جلب الصفحة {page_number}: {e}")
        break # إيقاف الحلقة في حال وجود خطأ في الشبكة

    soup = BeautifulSoup(response.text, "html.parser")
    products_container = soup.find("ol", class_="row")

    # التحقق من وجود حاوية المنتجات لتجنب الأخطاء في الصفحات الفارغة
    if not products_container:
        print("  -> لم يتم العثور على منتجات في هذه الصفحة.")
        continue

    products = products_container.find_all("li", class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")

    for container in products:
        box = container.find("article", class_="product_pod")
        if not box:
            continue

        # اسم المنتج
        name = box.select_one("h3 a")["title"]

        # رابط المنتج (يُفضل استخدام root_url لضمان صحة الرابط)
        product_url = urljoin(root_url, box.select_one("h3 a")["href"])

        # رابط الصورة
        image_url = urljoin(root_url, box.select_one(".image_container img")["src"])

        # السعر
        price = box.select_one(".price_color").get_text(strip=True)

        # التقييم
        rating_element = box.select_one("p.star-rating")
        rating_name = rating_element.get("class")[1] if rating_element and len(rating_element.get("class")) > 1 else "Zero"
        rating = ratings.get(rating_name, 0)

        # التوفر
        availability = box.select_one(".availability").get_text(" ", strip=True)

        # 3. إضافة البيانات للقائمة الموحدة (وليس إنشاء قائمة جديدة)
        scraped_data.append({
            "الاسم": name,
            "السعر": price,
            "التقييم": rating,
            "التوفر": availability,
            "الرابط": product_url,
            "الصورة": image_url
        })

    # تأخير بسيط لتجنب حظر الطلبات المتتالية السريعة
    time.sleep(1)

# 4. حفظ البيانات في CSV مرة واحدة فقط بعد انتهاء الحلقة بالكامل
csv_filename = "my_data.csv"
with open(csv_filename, mode="w", encoding="utf-8-sig", newline="") as file:
    fieldnames = ["الاسم", "السعر", "التقييم", "التوفر", "الرابط", "الصورة"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(scraped_data)

print("=" * 50)
print(f" تم استخراج {len(scraped_data)} منتج بنجاح من جميع الصفحات!")
print(f" تم حفظ البيانات في الملف: {csv_filename}")
```
